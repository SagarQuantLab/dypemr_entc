**8. Operators in python**
- Arithmetic Operators          `+, -, *, /, %`
- Relational Operators          `<, <=, >, >=, ==, !=`
- Logical Operators             `and, or, not`
- Bitwise Operator              `&, |, ^, ~, <<, >>`
- Assignment Operator           `=, +=, -=, *=, %=`
- Indentity Operator            `is, is not`

```python
#8.1 Arithmetic Operators
first_num = 10
second_num  = 11

# addition
print("Addition:", first_num + second_num)  

# subtraction
print("Subtraction:", first_num - second_num) 

# multiplication
print("Multiplication:", first_num * second_num)  

# division
print("Division:", first_num / second_num) 

# floor Division
print("Floor Division:", first_num // second_num)  

# modulus
print("Modulus:", first_num % second_num) 

# exponentiation
print("Exponentiation:", first_num ** second_num)
```

```python
#8.2 Comparision Operator
first_num = 10
second_num = 5

print(first_num > second_num)
print(first_num < second_num)
print(first_num == second_num)
print(first_num != second_num)
print(first_num >= second_num)
print(first_num <= second_num)
```

```python
#8.3 Logical operator
first_num = True
second_num = False

print(first_num and second_num)
print(first_num or second_num)
print(not first_num)
```

```python
# 8.4 Bitwise operator works for on binary number
first_num = 10
second_num = 5

print(first_num & second_num)
print(first_num | second_num)
print(~first_num)
print(first_num ^ second_num)
print(first_num >> 2)
print(first_num << 2)
```

```python
# 8.5 Assignment Operators
first_num = 10
second_num = first_num

print(second_num)
second_num += first_num
print(second_num)
second_num -= first_num
print(second_num)
second_num *= first_num
print(second_num)
second_num <<= first_num
print(second_num)
```

```python
# 8.6 Identity Operator
first_num = 10
second_num = 5
third_num = first_num

print(first_num is not second_num)
print(first_num is second_num)
```

```python
# 9. strings and their operations
my_string = "This is COEP College"

print(f"first letter - {my_string[0]}")
print(f"second letter - {my_string[1]}")
print(f"last letter - {my_string[-1]}")
print(f"second last letter - {my_string[-2]}")
print(f"first 3 letter of string  - {my_string[0:3]}")
print(f"interval of 2 - {my_string[0::2]}")
print(f"reverse string = {my_string[::-1]}")
print(f"last 3 words - {my_string[:len(my_string)-4:-1]}")
print(f"reverse with interval of 2 - {my_string[::-2]}")
print(f"Upper case - {my_string.upper()}")
print(f"Lower case - {my_string.lower()}")
print(f"T 10 times - {my_string[0]*10}")
print(f"add in string - {my_string + " Hello"} ")
print(f"replace first letter - {my_string}")
print(f"replace - {my_string.replace("T", "Z")}")
print(f"count - {my_string.count("T")}")
```

```python
# list and their operations
my_list = [
    "Apple", "Mango", "Banana"
    ]

my_list_new = [
    "Kiwi", "Guava"
]


print(f"first item - {my_list[0]}")
print(f"last item - {my_list[-1]}")
print(f"sorted - {my_list.sort()}")
print(f"Replace - {my_string.replace("Orange", "Mausambi")}")
my_list.append("Orange")
print(f"append list - {my_list}")
my_list.append(my_list_new)
print(my_list)

my_list = [
    "Apple", "Mango", "Banana"
    ]

my_list_new = [
    "Kiwi", "Guava"
]
my_list.extend(my_list_new)
print(my_list)

print(f"remove last element - {my_list.pop()}")
my_list.insert(5, "Guava")
my_list[-1] = "Carrot"
print(my_list)
print(f"clear list - {my_list.clear()}")
```

```python
# Dictionary
my_dict = {
    "Name" : "Rohan",
    "Age" : 35,
    "Gender" : "M"
}

print(f"keys - {my_dict.keys()}")
print(f"values - {my_dict.values()}")
my_dict["Religion"] = "Hindu"
print(f"add entry - {my_dict}")
print(f"get all things - {my_dict.items()}")
my_dict.pop("Gender")
print(my_dict)
my_dict["Age"] = 16
print(my_dict)
my_dict.update({"Gender" : "M"})
print(my_dict)
```

```python
# sets
my_set = {"apple", "banana", "cherry"}

print(my_set) # changes the order any time
my_set.add("orange")
print(my_set)

tropical = {"pineapple", "mango", "papaya"}
my_set.update(tropical)

my_set.remove("mango") # this will raise error is mango not found
my_set.discard("mango") #  this will not raise error when not found
my_set.pop() # this will remove random entry
```
